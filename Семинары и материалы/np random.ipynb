```python
import numpy as np
seed, n, low, high = map(int, input().split())
np.random.seed(seed)
a = np.random.uniform(size=n, low=low, high=high)
print(' '.join(map(lambda x: f'{x:.2f}', [a.mean(), a.std()])))
```

```python
seed, t, n = map(int, input().split())
np.random.seed(seed)
a = np.random.poisson(lam=t, size=n)
print(' '.join(map(lambda x: f'{x:.2f}', [a.mean(), a.std()])))
```

```python
seed, n, p, m = input().split()
seed, n, m = map(int, [seed, n, m])
p = float(p)
np.random.seed(seed)
diff = []
for i in range(m + 1):
    diff.append(n)
    n = round(n * np.random.binomial(1, p, n).mean())
diff = [abs(diff[i] - diff[i - 1]) / diff[i - 1] for i in range(1, m + 1)]
print(' '.join(map(lambda x: f'{x:.3f}', diff)))
```
